# Python demo

This notebook runs in the browser with `xeus-python`. NumPy and Matplotlib are included in the WebAssembly environment.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)
print(f"Computed {y.size} points; mean={y.mean():.3f}")
plt.plot(x, y, label="sin(x)")
plt.title("A sine wave")
plt.xlabel("x")
plt.legend()
plt.show()
```
